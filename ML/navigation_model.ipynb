# Step 1: Introduction

Machine learning is used to recommend places instead of only displaying them.

The system suggests a location based on time and user needs.


# Step 2: Import Libraries

We import the Decision Tree model from sklearn.


```python
from sklearn.tree import DecisionTreeClassifier
```

# Step 3: Prepare Data

We create a dataset with:
- Time of day (hour)
- Category (food, fitness, health)

These will be used as input features.


```python
# Input:
# category: food=0, fitness=1, health=2

X = [
    [9, 2],
    [12, 0],
    [18, 1],
    [20, 0],
    [7, 2],
    [17, 1],
    [13, 0]
]

y = [
    "ITTI Health Clinic",
    "ITTI Pizza",
    "ITTI Gym",
    "ITTI Grill",
    "ITTI Pharmacy",
    "ITTI Gym",
    "ITTI Pizza"
]
```

# Step 4: Train the Model

We train the Decision Tree model using the dataset.


```python
model = DecisionTreeClassifier()
model.fit(X, y)
```

# Step 5: Make Prediction

We test the model with a new input.


```python
# Example: 16:00, fitness
prediction = model.predict([[16, 1]])

print("Recommended Place:", prediction[0])
```

# Step 6: Conclusion

The model recommends places based on user input, making the system more interactive.
