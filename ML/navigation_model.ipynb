# ITTI Smart Town App
## Machine Learning for Smart Navigation

This notebook demonstrates a simple machine learning model that predicts
which location in Iitti a user might want to visit based on the time of day.
A Decision Tree classifier is used for this purpose.

```python
import numpy as np
from sklearn.tree import DecisionTreeClassifier
```

```python
X = [
    [9],
    [10],
    [14],
    [17],
    [20]
]

y = [
    "Library",
    "Library",
    "Park",
    "Grocery Store",
    "Park"
]

model = DecisionTreeClassifier()
model.fit(X, y)
```

The dataset represents typical popular places in Iitti at different hours of the day.

```python
prediction = model.predict([[16]])

print("Recommended location:", prediction[0])
```

Recommended location: Grocery Store
