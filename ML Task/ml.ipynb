## 1. Introduction

In this task, I analyze the population structure of Finland based on age groups.

The goal is to understand how different age groups have changed over time and what this means for the future.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

## 2. Data Loading

The dataset contains population values divided into three age groups:

- 0–14 (young population)
- 15–64 (working age population)
- 65+ (elderly population)

```python
df = pd.read_csv("finland_age_population.csv")
print(df.head())
```

## 3. Data Cleaning

The dataset was already structured properly, so only basic checks were performed.

No missing values were found.

```python
df = df.dropna()
```

## 4. Population by Age Group

The following graph shows how each age group has changed over time.

```python
plt.plot(df["Year"], df["Age_0_14"], label="Age 0-14")
plt.plot(df["Year"], df["Age_15_64"], label="Age 15-64")
plt.plot(df["Year"], df["Age_65_plus"], label="Age 65+")

plt.title("Population by Age Group in Finland")
plt.xlabel("Year")
plt.ylabel("Population")
plt.legend()
plt.show()
```

## 5. Analysis

The graph shows that the number of elderly people (65+) is increasing steadily.

At the same time, the number of young people (0–14) is decreasing.

This indicates that Finland's population is aging.

## 6. Turning Points

After around 2010, the growth of the elderly population becomes more noticeable.

This suggests that aging is becoming a major demographic trend.

## 7. Prediction Model

A linear regression model was used to predict future population trends for different age groups.

The model uses the year as the main variable.

```python
X = df[["Year"]]

model_young = LinearRegression()
model_young.fit(X, df["Age_0_14"])

model_old = LinearRegression()
model_old.fit(X, df["Age_65_plus"])
```

## 8. Future Prediction

The model is used to estimate how age groups will change in the future.

```python
future = pd.DataFrame({"Year": [2030, 2040]})

pred_young = model_young.predict(future)
pred_old = model_old.predict(future)

print("Young population prediction:", pred_young)
print("Elderly population prediction:", pred_old)
```

## 9. Visualization of Prediction

The predictions are added to the graph to show future trends.

```python
plt.scatter(df["Year"], df["Age_0_14"], label="Young (real)")
plt.scatter(df["Year"], df["Age_65_plus"], label="Elderly (real)")

plt.scatter(future["Year"], pred_young, label="Young (predicted)")
plt.scatter(future["Year"], pred_old, label="Elderly (predicted)")

plt.legend()
plt.title("Future Age Group Trends")
plt.xlabel("Year")
plt.ylabel("Population")
plt.show()
```
